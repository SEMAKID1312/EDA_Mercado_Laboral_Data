### Limpieza

```python
# Duplicados
df_data.duplicated().sum()
```

```python
df_data.drop_duplicates(inplace=True)
df_data.duplicated().sum()
```

```python
# Eliminar columnas constantes (una sola categoría)
cols_constantes = [c for c in df_data.columns if df_data[c].nunique() == 1]
df_data.drop(columns=cols_constantes, inplace=True)
```

```python
# Missings por columna
df_data.isnull().sum().sort_values(ascending=False)
```

```python
serie_nulos = df_data.isna().mean() * 100
serie_nulos
```

La columna JobSat, se deja como NaN, debido a que es una respuesta valida. Además, de ser una variable que interfiere directamente en las hipótesis planteadas y contiene demasiados valores NaN como para eliminarlos, se toma la decisión de mantenerlo tal y como está.

Luego, la columna más importante del dataset tiene un 43.23% de valores nulos, se podría sustituir por la media según el país y el tipo de desarrollador, sin embargo es una información que sería sesgada finalmente y no enriquecería el estudio, y como existe un margen de datos relativamente amplio, es por lo que se decide eliminar aquellas filas nulas que tan importantes son para el estudio.

```python
# Eliminar filas con valores nulos en "ConvertedCompYearly"
df_data.dropna(subset=["ConvertedCompYearly"], inplace=True)
```

Aplicamos mismo método para EdLevel, debido a que son pocas filas las que tienen valor nulo, no interfiere en el estudio.

```python
df_data = df_data.dropna(subset=["EdLevel"])
```

RemoteWork realmente no aporta nada importante para el estudio, se elimina la columna

```python
# Eliminar columna RemoteWork
df_data.drop(columns=["RemoteWork"], inplace= True)
```

Lo mismo ocurre con el lenguaje de programación, no aportaría nada para este estudio en concreto, se elimina

```python
df_data.drop(columns=["LanguageHaveWorkedWith"], inplace= True)
```

Los datos de la columna AISelect pueden resumirse en 3; Sí, No, NC (No Contesta)

```python
df_data["AISelect"].value_counts()
```

```python
mapeo_ai = {
    "Yes, I use AI tools daily": "Yes",
    "Yes, I use AI tools weekly": "Yes",
    "Yes, I use AI tools monthly or infrequently": "Yes",
    "No, and I don't plan to": "No",
    "No, but I plan to soon": "No"
}

df_data["AISelect"] = df_data["AISelect"].replace(mapeo_ai).fillna("NA") # NA : No Answer
```

```python
mapeo_edlevel = {
    "Bachelor’s degree (B.A., B.S., B.Eng., etc.)": "Bachelor",
    "Professional degree (JD, MD, Ph.D, Ed.D, etc.)": "Professional",
    "Master’s degree (M.A., M.S., M.Eng., MBA, etc.)": "Master",
    "Some college/university study without earning a degree": "Some College",
    "Associate degree (A.A., A.S., etc.)": "Associate",
    "Secondary school (e.g. American high school, German Realschule or Gymnasium, etc.)": "Secondary",
    "Primary/elementary school": "Primary",
    "Other (please specify:)": "Other"
}

df_data["EdLevel"] = df_data["EdLevel"].replace(mapeo_edlevel)
```

```python
mapeo_employment = {                                            
    'Freelancer':'Freelancer'
    }

df_data["Employment"] = df_data["Employment"].replace(mapeo_employment)
```

```python
mapeo_devtype = {                                            
    "Data or business analyst":"Data analyst",
    "Database administrator or engineer":"Database administrator",    
    "Developer, AI apps or physical AI":"AI developer"
    }

df_data["DevType"] = df_data["DevType"].replace(mapeo_devtype)
```

```python
mapeo_country = {                                            
    'United Kingdom of Great Britain and Northern Ireland':'UK and Ireland',
    'United States of America':'USA',
    'Iran, Islamic Republic of...':'Iran',
    'Hong Kong (S.A.R.)':'Hong Kong',
    'United Arab Emirates':'UAE',
    'Russian Federation':'Russia',
    "Côte d'Ivoire":'Ivory Coast',
    'Congo, Republic of the...':'Congo'
    }

df_data["Country"] = df_data["Country"].replace(mapeo_country)
```

Las tres columnas que quedan, se aplicará la moda y media, respectivamente. Antes se indica en una nueva columna que era un valor nulo en esa columna y fila.

```python
# Marcar antes de imputar
df_data["fue_missing_YearCode"] = df_data["YearsCode"].isna()
df_data["fue_missing_WorkExp"] = df_data["WorkExp"].isna()
df_data["fue_missing_Industry"] = df_data["Industry"].isna()
```

```python
df_data["Industry"] = df_data["Industry"].fillna(df_data["Industry"].mode()[0])
```

```python
df_data["WorkExp"] = df_data["WorkExp"].fillna(df_data["WorkExp"].median())
```

```python
df_data["YearsCode"] = df_data["YearsCode"].fillna(df_data["YearsCode"].median())
```

Desde el análisis univariante se ha comprobado que existen outliers, de los cuales el valor 100 se descarta. 

```python
df_data["YearsCode"].unique()
```

```python
df_data.loc[df_data["YearsCode"] == 100, "YearsCode"] = pd.NA
```

```python
df_data = df_data.dropna(subset=["YearsCode"])
```
